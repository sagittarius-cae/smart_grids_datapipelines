```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_date
# Import tool for clean jupyter
from IPython.display import clear_output
```

# Initialize Spark with Iceberg pointing to our clean data lakehouse mount

```python
spark_energy = SparkSession.builder \
    .appName("CleanMedallionIngestion") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.5.2") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.local.type", "hadoop") \
    .config("spark.sql.catalog.local.warehouse", "/app/data_lakehouse/") \
    .config("spark.ui.enabled", "false") \
    .getOrCreate()
```

```python
spark_energy.sparkContext.setLogLevel("ERROR")
spark_energy.sql("USE local")
```
