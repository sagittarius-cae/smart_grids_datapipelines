```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, current_timestamp, trim, upper, to_date

# Initialize Spark with Iceberg extensions & local warehouse configuration
spark = SparkSession.builder \
    .appName("CSVToSparkIcebergProcessing") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.5.0") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.local.type", "hadoop") \
    .config("spark.sql.catalog.local.warehouse", "./data_lakehouse") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
print("Spark session active! Version:", spark.version)
```

```python
# Read CSV with headers enabled and automatic schema inference
global_generation_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("./data_lakehouse/release_generation_yearly_global.csv")  # Replace with your actual CSV path

# Inspect raw schema and initial rows
global_generation_df.printSchema()
global_generation_df.show(5)
```

```python
spark.sql("USE local")
spark.sql("CREATE DATABASE IF NOT EXISTS db_raw.energy_generations")
```

```python
global_generation_df.writeTo("local.db_raw.energy_generations") \
    .tableProperty("format-version", "2") \
    .createOrReplace()

print("✅ ¡Tabla creada exitosamente en `./data_lakehouse/db_raw.db/energy_generations`!")
```

```python
global_generations_rawdata = spark.read \
    .format("iceberg") \
    .load("local.db_raw.energy_generations")

global_generations_rawdata.show(5)
```
